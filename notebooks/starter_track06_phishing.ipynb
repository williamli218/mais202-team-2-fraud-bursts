**Why hostnames.** Every legitimate URL in the raw file is written as a bare `https://www.<domain>` homepage and most phishing URLs aren't. A model on full URLs learns the formatting and nothing about phishing. So the loader cuts every URL down to its hostname, collapses duplicates, drops hostnames that appear with both labels, and holds out a stratified 15% as the test set. Don't undo any of that. The raw label is 1 for legitimate; the loader flips it so `is_phishing == 1` means phishing.

## 8. Timing a single prediction

The second number in your metric is the 95th-percentile time to score one
link, because this decision happens while the user waits. Measure it on
single calls to `predict()`, not on a batch: a batch hides the per-call
overhead that dominates at one link at a time. The pattern:

```python
import time
times = []
for url in sample_urls:
    t0 = time.perf_counter(); predict(url); times.append(time.perf_counter() - t0)
metrics.p95_latency_ms(times)
```
